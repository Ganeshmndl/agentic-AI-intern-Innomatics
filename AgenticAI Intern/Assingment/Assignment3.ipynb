```python
# Employee performance scores
employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

# Finding the highest performance score
highest_score = max(employees.values())

# Finding employees with the highest score
top_performers = []
for name, score in employees.items():
    if score == highest_score:
        top_performers.append(name)

# Printing result
print("Top Performers Eligible for Bonus:", ", ".join(top_performers), f"(Score: {highest_score})")
```

```python
# Input search query
query = "Buy mobile phone buy phone online"

# Converting to lowercase
query = query.lower()

# Removing basic punctuation (if any)
query = query.replace(".", "").replace(",", "")

# Splitting into words
words = query.split()

# Dictionary to store keyword frequency
keyword_count = {}

# Counting frequency of each keyword
for word in words:
    if word in keyword_count:
        keyword_count[word] += 1
    else:
        keyword_count[word] = 1

# Displaying keywords searched more than once
result = {}
for key, value in keyword_count.items():
    if value > 1:
        result[key] = value

# Printing result
print(result)
```

```python
# Sensor readings where index represents hour
sensor_readings = [3, 4, 7, 8, 10, 12, 5]

# List to store valid sensor readings
valid_readings = []

# Checking each reading
for index, value in enumerate(sensor_readings):
    if value % 2 == 0:
        valid_readings.append((index, value))

# Printing result
print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)
```

```python
# List of employee email IDs
emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

# Dictionary to store domain counts
domain_count = {}

# Extracting domains and counting usage
for email in emails:
    domain = email.split("@")[1]
    if domain in domain_count:
        domain_count[domain] += 1
    else:
        domain_count[domain] = 1

# Calculating and printing percentage usage
total_emails = len(emails)
for domain, count in domain_count.items():
    percentage = (count / total_emails) * 100
    print(f"{domain}: {int(percentage)}%")
```

```python
# Daily sales data
sales = [1200, 1500, 900, 2200, 1400, 3000]

# Calculating average sales
total_sales = 0
for value in sales:
    total_sales += value

average_sales = total_sales / len(sales)

# Detecting sales spikes (more than 30% above average)
threshold = average_sales * 1.3

for day, sale in enumerate(sales, start=1):
    if sale > threshold:
        print(f"Day {day}: {sale}")
```

```python
# List of user IDs
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

# Dictionary to count user IDs
user_count = {}

# Counting occurrences
for user in user_ids:
    if user in user_count:
        user_count[user] += 1
    else:
        user_count[user] = 1

# Printing duplicate user IDs
for user, count in user_count.items():
    if count > 1:
        print(f"{user} → {count} times")
```

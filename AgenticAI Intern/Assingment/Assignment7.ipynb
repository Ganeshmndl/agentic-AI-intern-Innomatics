```python
# Smart Light
class SmartLight:
    
    def __init__(self, light_name):
        self.light_name = light_name
    
    def control_light(self, action):
        if action.upper() == "ON":
            print(f"{self.light_name} is ON")
        else:
            print(f"{self.light_name} is OFF")


# Sample input
light = SmartLight("Bedroom Light")
light.control_light("ON")
```

```python
# Employee ID Card
class Employee:
    
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department
    
    def display_id_card(self):
        print("Employee ID Card")
        print("Name:", self.name)
        print("ID:", self.emp_id)
        print("Department:", self.department)


# Sample input
employee = Employee("Rahul", "EMP102", "IT")
employee.display_id_card()
```

```python
# Mobile Contact
class Contact:

    # Constructor to store contact details
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

    # Method to display contact information
    def display_contact(self):
        print("Contact Saved")
        print("Name:", self.name)
        print("Phone:", self.phone)


# Sample input
contact = Contact("Anita", 9876543210)

# Display contact information
contact.display_contact()
```

```python
# Product Price Tag
class Product:

    # Constructor to store product details
    def __init__(self, name, price):
        self.name = name
        self.price = price

    # Method to display price tag
    def display_price_tag(self):
        print("Product:", self.name)
        print("Price: ₹" + str(self.price))


# Sample input
product = Product("Headphones", 2499)

# Display price tag
product.display_price_tag()
```

```python
# Movie Rating
class Movie:

    # Constructor to store movie details
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating

    # Method to display movie details
    def display_movie(self):
        print("Movie:", self.name)
        print("Rating:", self.rating)


# Sample input
movie = Movie("Inception", 4.8)

# Display movie details
movie.display_movie()
```
